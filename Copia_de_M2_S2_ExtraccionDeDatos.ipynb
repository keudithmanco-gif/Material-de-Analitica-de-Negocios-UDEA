# <font color='056938'> **Extraccion de datos** </font>

La extracción de datos es el primer paso en el proceso ETL (Extract, Transform, Load). Consiste en reunir datos de uno o varios sistemas fuente para que puedan ser transformados y cargados en un almacén de datos objetivo o un repositorio de datos. Los sistemas fuente pueden ser **bases de datos, archivos planos, APIs, servicios web, registros (logs) u cualquier otro repositorio de datos**. Los datos extraídos pueden ser crudos y sin procesar.

La extracción de datos es un paso crítico en el proceso ETL porque la calidad y precisión de los datos extraídos afectan la calidad de las ideas y los informes generados posteriormente. El proceso de extracción se puede programar para que ocurra a intervalos regulares (por ejemplo, diariamente, por hora) para garantizar que los datos en el sistema objetivo estén actualizados.

Existen diversos métodos para la extracción de datos, que incluyen:

- **Extracción completa**: Se extraen todos los datos del sistema fuente en su totalidad. Esto se hace generalmente durante la configuración inicial o cuando los datos del sistema fuente cambian con poca frecuencia.

- **Extracción incremental**: Se extraen solo los datos nuevos o modificados desde la última extracción. Este método es más eficiente para conjuntos de datos grandes y actualizaciones frecuentes.

- **Captura de Cambios en los Datos (CDC, por sus siglas en inglés)**: La CDC es una técnica utilizada para identificar y capturar cambios en los datos fuente desde la última extracción. Es particularmente útil para la integración de datos en tiempo real o casi en tiempo real.

- **Instantánea Periódica**: Los datos se extraen a intervalos específicos, independientemente de los cambios, para crear una instantánea histórica.

La elección del método de extracción depende de factores como el volumen de datos, la frecuencia de actualización y los requisitos específicos del proceso ETL.

Considere el caso en el cual tenemos un conjunto de solicitudes de servicio en distintas partes de la ciudad país (cada una con su geolocalización)

| Servicio                                  | Latitud    | Longitud    |
|----------------------------------------|-------------|--------------|
| Servicio 1| 6.25264   | -75.56914   |

Queremos construir un archivo de datos con la distancia entre cada par de solucitudes para con ello programar la ruta.

Para ello, usaremos la siguiente función a la cual le entregamos la latitud y longitud de dos sitios y retorna la distancia entre ellos haciendo uso del servicio web expuesto por el proyecto [Open Source Routing Machine](https://project-osrm.org/)

---

OSRM es un **motor de enrutamiento de código abierto** que permite calcular rutas, distancias y tiempos de viaje en la red vial.

* Utiliza datos de **OpenStreetMap** (OSM).
* Expone un **API HTTP** (servicio web) que se puede consultar con peticiones GET.
* Devuelve la información en formato **JSON**.


Usaremos el siguiente `endpoint`, es decir una dirección o URL específica del  servicio web a la que se envían solicitudes y desde donde se reciben respuestas:

```python
http://router.project-osrm.org/route/v1/driving/{origen};{destino}
```
En esta dirección:

* `route/v1/` → indica que es el servicio de **ruteo** versión 1.
* `driving` → indica el perfil de transporte (otros posibles: `walking`, `cycling`).
* `{origen};{destino}` → coordenadas en el formato `longitud,latitud`.

Por ejemplo, para calcular la distancia entre `(-74.08175,4.60971)` y `(-73.93583,4.66707)` tendríamos:

```
http://router.project-osrm.org/route/v1/driving/-74.08175,4.60971;-73.93583,4.66707
```


El servicio devuelve un JSON con varios campos, por ejemplo:

```json
{
  "routes": [
    {
      "distance": 15732.1,
      "duration": 1123.4,
      "legs": [
        {
          "distance": 15732.1,
          "duration": 1123.4,
          "steps": [...]
        }
      ]
    }
  ],
  "waypoints": [...]
}

Dónde se identifica:
```

* `routes[0].legs[0].distance` → distancia en **metros**.
* `routes[0].legs[0].duration` → duración en **segundos**.

---

Definamos la función `calcular_distancia()` que hace uso de este servicio para calcular la distancia entre dos de los puntos que recibimos en nuestro listado

````python
import requests, json
import pandas as pd

def calcular_distancia (punto1: tuple, punto2: tuple):
  """
    Calculate the driving distance between two points using the OSRM API.

    Args:
        punto1 (tuple): A tuple containing the latitude and longitude of the first point.
        punto2 (tuple): A tuple containing the latitude and longitude of the second point.

    Returns:
        float: The driving duration in seconds between the two points.
  """

  url = 'http://router.project-osrm.org/route/v1/driving/'
  o1 = str(str(punto1[1]) +',' + str(punto1[2]))
  o2 = str(str(punto2[1]) +',' + str(punto2[2]))
  x = o1 + ';' + o2
  response = requests.get(url+x)
  data = json.loads(response.content)

  return data['routes'][0]['legs'][0]['distance']


````

Probemos la función con dos puntos específicos:


````python
data = calcular_distancia(("Servicio 1", 13.388860, 52.517037), ("Servicio 2", 13.397634, 52.529407))
data
````

Calculemos la distancia para todos los pares de puntos en una lista de puntos y craremos un dataframe con ello

````python
# Lista de sitios
sitios = [
    ("Servicio 1", 13.388860, 52.517037),
    ("Servicio 2", 13.397634, 52.529407),
    ("Servicio 3", 13.376760, 52.503770),
    ("Servicio 4", 13.407674, 52.439097),
]

# Crea una lista vacia de distancias
distancias = []

#Calcula las distancias entre cada par
for i in range(len(sitios)):
    for j in range(i + 1, len(sitios)):
      dist = calcular_distancia(sitios[i], sitios[j])
      distancias.append([sitios[i][0],  sitios[j][0], dist])

distancias
````

ahora creamos un dataframe con la información obtenida

````python
# Create a DataFrame from the list of lists
df = pd.DataFrame(distancias, columns=['origen', 'destino', 'distancia'])
df
````

## <font color='8EC044'> **Consolidación de datos** </font>



Una vez hemos obtenido los datos de distintas fuentes, una primera tarea que quisieramos es poder eliminar o adicionar registros a los dataframes que hemos construido. Para ello usamos funciones como las siguientes:

| Método                   | Descripción                                                                                     |
|--------------------------|-------------------------------------------------------------------------------------------------|
| `df.loc[index] = data`   | Agrega una fila específica utilizando una etiqueta de índice y datos proporcionados en `data`.   |
| `df.loc[len(df)] = data` | Agrega una fila al final del DataFrame utilizando la longitud actual del DataFrame como índice. |
| `df.iloc[index] = data`  | Agrega una fila específica utilizando un índice numérico y datos proporcionados en `data`.      |
| `df.loc[len(df)] = data` | Agrega una fila al final del DataFrame utilizando la longitud actual del DataFrame como índice. |



De forma similar, pueden eliminarse filas o columnas del dataframe:

| Método                        | Descripción                                                                                             |
|-------------------------------|---------------------------------------------------------------------------------------------------------|
| `df.drop(index)`              | Elimina una fila específica utilizando una etiqueta de índice.                                          |
| `df.drop(columns)`            | Elimina columnas específicas del DataFrame.                                                              |
| `df.drop_duplicates()`        | Elimina filas duplicadas basadas en todas las columnas o columnas específicas.                         |
| `df.drop(labels, axis=1)`    | Elimina columnas específicas utilizando una lista de etiquetas.                                          |


Sin embargo, en la mayoria de los casos estamos interesados no en agregar filas una a una, sino en concatenar datos de diferentes dataframes. Para lo cual Pandas ofrece diferentes opciones dependiendo del tipo de unión que deseemos hacer.

| Tipo de unión | Función de Pandas | Descripción |
| --- | --- | --- |
| Unión interna | `pd.merge(df1, df2, on=columna_común)` | Retorna sólo las filas en las que existe una coincidencia entre ambas tablas en la columna especificada. |
| Unión externa | `pd.merge(df1, df2, on=columna_común, how='outer')` | Retorna todas las filas de ambas tablas, llenando con NaNs las celdas sin coincidencia. |
| Unión izquierda | `pd.merge(df1, df2, on=columna_común, how='left')` | Retorna todas las filas de la tabla izquierda, y las filas de la tabla derecha que coinciden con la columna especificada. Las celdas sin coincidencia se llenan con NaNs. |
| Unión derecha | `pd.merge(df1, df2, on=columna_común, how='right')` | Retorna todas las filas de la tabla derecha, y las filas de la tabla izquierda que coinciden con la columna especificada. Las celdas sin coincidencia se llenan con NaNs. |
| Concatenación | `pd.concat([df1, df2])` | Une dos o más DataFrames en función de las filas o columnas. |

Considere por ejemplo dos dataframes diferentes

````python
import pandas as pd

df1 = pd.DataFrame({'A': ['A0', 'A1', 'A2', 'A3'],
                    'B': ['B0', 'B1', 'B2', 'B3'],
                    'C': ['C0', 'C1', 'C2', 'C3'],
                    'D': ['D0', 'D1', 'D2', 'D3']})

print(df1)
````

````python
df2 = pd.DataFrame({'A': ['A4', 'A5', 'A6', 'A7'],
                    'B': ['B4', 'B5', 'B6', 'B7'],
                    'C': ['C4', 'C5', 'C6', 'C7'],
                    'D': ['D4', 'D5', 'D6', 'D7']})

print(df2)
````

La siguiente instrucción concatena las filas de los dos DataFrames

````python
df_concatenado = pd.concat([df1, df2], axis=0, ignore_index=True)
df_concatenado
````

La opción `axis=0` indica se agregan las filas, mientras que la opción `ignore_index=True` indica que se crea un nuevo conjunto de indices para el DataFrameConsolidado
