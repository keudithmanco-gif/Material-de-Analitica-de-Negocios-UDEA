Uno podria estar interesado en seleccionar un subconjunto de una tabla o dataframe. Esto puede implicar seleccionar filas, columnas o ambas. Aunque existen diversas opciones para hacerlo, veremos aquí a modo de ejemplo solo un par de ellas:


### <font color='8EC044'> **Selección de filas/registros** </font>

Existen distintas formas de seleccionar filas, algunos de ellos son:

| Método | Descripción | Ejemplo |

# <font color='056938'> **Crear nuevas columnas** </font>


Una de las tareas más comunes en los dataframes es generar nuevas columnas con base en el dataframe construído. Creemos por ejemplo la columna que indique que porcentaje de la capacidad total del sistema representa la capacidad de cada estación

```python
df_info['% capacity'] = df_info['capacity']/df_info['capacity'].sum()
df_info.head()
```

Algunos calculos son más complejos y debemos aplicar funciones que hemos definido previamente. Para ello usamos la función `apply()`

Considere el caso en el que deseamos tomar la primera y última parte del código de identificación externo

> **3ea89109**-d2f3-46eb-9c41-**c43742050340**

Creamos la función `acortar_codigo()` que toma las dos partes requeridas del código.

```python
def acortar_codigo(codigo: str) -> str:
    """
    Dado un código con formato UUID (separado por guiones),
    retorna una versión más corta formada por la primera
    y la última parte del código.

    Ejemplo:
    3ea89109-d2f3-46eb-9c41-c43742050340
    -> 3ea89109-c43742050340
    """
    partes = codigo.split("-")
    if len(partes) < 2:
        return codigo  # devuelve el original si no tiene el formato esperado
    return f"{partes[0]}-{partes[-1]}"


# --- Ejemplo de uso ---
original = "3ea89109-d2f3-46eb-9c41-c43742050340"
print(acortar_codigo(original))

```

Ahora aplicamos la función a la columna `external_id` del DataFrame para obtener la nueva columna:

```python
df_info["shorthen_id"] = df_info["external_id"].apply(acortar_codigo)
df_info.head()
```

Note que es posible que la nueva columna involucre valores de diferentes columna del mismo registro, para ello requerimos usar toda la fila asociada al registro en el dataframe.

Considere el caso en el que queremos dientificar un café cercano a cada una de las estaciones. Para este tipo de operaciones más complejas podemos  usar funciones personalizadas que creamos y las aplicamos al DataFrame a través de la función `apply()`.

En este caso, definimos la función `buscar_cafe_cercano()` que dadas unas coordenadas busca el café más cercano.

```python
import requests

def buscar_cafe_cercano(row) -> str:
    """
    Busca el nombre de un café cercano a las coordenadas dadas usando Overpass API.

    Args:
        lat (float): Latitud del punto.
        lon (float): Longitud del punto.

    Returns:
        str: Nombre del café encontrado o un mensaje si no hay cafés cerca.
    """
    overpass_url = "http://overpass-api.de/api/interpreter"

    # Query Overpass: buscar cafés en 500m alrededor
    query = f"""
    [out:json];
    node["amenity"="cafe"](around:500,{row['lat']},{row['lon']});
    out;
    """

    try:
        response = requests.get(overpass_url, params={'data': query})
        data = response.json()

        if "elements" in data and len(data["elements"]) > 0:
            # Tomamos el primer café
            cafe = data["elements"][0]
            name = cafe.get("tags", {}).get("name", "Café sin nombre")
            return name
        else:
            return "No se encontró ningún café cerca."
    except Exception as e:
        return f"Error consultando Overpass: {e}"


# --- Ejemplo: UdeA ---
print(buscar_cafe_cercano(df_info.iloc[8]))
```

Aplicamos la función a nuestro DataFrame para calcular la nueva columna

```python
selection = df_info.sample(10)
selection['coffe_shop'] = selection.apply(buscar_cafe_cercano, axis = 1)

selection.head()
```

# <font color='056938'> **Wide format vs long format** </font>

## <font color='8EC044'> **Wide Format (Formato Amplio):**
   
   En el formato amplio, cada variable se representa en una columna diferente del DataFrame. Esto significa que cada columna contiene una variable diferente, y cada fila representa una observación o un conjunto de observaciones relacionadas. El formato amplio es ideal para conjuntos de datos donde se tienen pocas variables y se desea una representación compacta.

   
   El siguiente es un ejemplo de un DataFrame en formato amplio:


```python
import pandas as pd

!gdown 1Lz127XdrCG2eOX4Bmokv4yEIqxLka-GR

df_wide = pd.read_excel('wide_format.xlsx')
df_wide.head()
```

En este ejemplo, cada columna (`cleaning`	`preventive`	`corrective`	`maintenance`) representa una variable diferente, y cada fila contiene observaciones relacionadas.


## <font color='8EC044'> **Long Format (Formato Largo):**

En el formato largo, se almacenan múltiples observaciones de diferentes variables en filas individuales. Esto significa que hay una columna adicional que indica la variable a la que pertenece cada observación. El formato largo es útil cuando se necesita realizar análisis y visualizaciones que requieren desglosar y comparar valores de diferentes variables.

La información que anteriormente vimos en formato amplio se veria de la siguiente forma en el formato largo

```python
df_long = pd.melt(
    df_wide,
    id_vars=["station_id"],
    value_vars=["cleaning", "preventive", "corrective", "maintenance"],
    var_name="service",
    value_name="date"
)

df_long.sample(10)
```


El formato amplio es más compacto y adecuado para conjuntos de datos simples, mientras que el formato largo es más flexible y útil cuando necesitas realizar análisis detallados o visualizaciones complejas que involucran múltiples variables.

Pandas proporciona funciones como `melt()` y `pivot()` para convertir datos entre estos dos formatos según tus necesidades.
