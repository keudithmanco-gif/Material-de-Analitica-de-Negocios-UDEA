## <font color='8EC044'> **Bases de datos relacionales `RDB`** </font>



Una **base de datos relacional** (`RDB`) es una forma de estructurar información en tablas, filas y columnas. Un RDB tiene la capacidad de establecer vínculos (o relaciones) entre información mediante el relacionameinto entre tablas, lo que facilita la comprensión y la obtención de estadísticas sobre la relación entre varios datos. Las tablas se vinculan entre sí por un campo en común, en ambos casos posee las mismas características como por ejemplo el nombre de campo, tipo y longitud; a este campo generalmente se le denomina ID, identificador o clave.







Un **esquema de base de datos relacional** es un plano o estructura que define cómo se organiza la información dentro de una base de datos relacional. Describe las tablas, los campos (o columnas) dentro de esas tablas, las relaciones entre tablas y las restricciones aplicadas a los datos. El esquema sirve como un marco para cómo se almacena, accede y gestiona la información en la base de datos. sus componentes claves son:



* **Tablas**: Los bloques de construcción fundamentales de un esquema. Cada tabla representa una entidad (por ejemplo, `Clientes`, `Pedidos`) y consta de filas (registros) y columnas (atributos).

* **Columnas**: Cada tabla tiene un conjunto de columnas que definen los atributos de la entidad. Cada columna tiene un tipo de dato específico (por ejemplo, `ENTERO`, `VARCHAR`, `FECHA`) que determina el tipo de datos que puede contener.

* **Claves Primarias**: Cada tabla incluye típicamente una clave primaria, que es un identificador único para cada registro en esa tabla. Esta clave asegura que no haya dos registros con el mismo valor en la(s) columna(s) de clave primaria.

* **Claves Foráneas**: Las claves foráneas se utilizan para establecer relaciones entre tablas. Una clave foránea en una tabla apunta a la clave primaria en otra tabla, creando un vínculo entre las dos tablas.

* **Relaciones**: El esquema define cómo se relacionan las tablas entre sí. Los tipos comunes de relaciones incluyen:
   - **Uno a Uno**: Un registro en una tabla corresponde a un solo registro en otra tabla.
   - **Uno a Muchos**: Un registro en una tabla puede relacionarse con múltiples registros en otra tabla.
   - **Muchos a Muchos**: Los registros en una tabla pueden relacionarse con múltiples registros en otra tabla, y viceversa (a menudo implementado mediante una tabla intermedia).

* **Restricciones**: Estas son reglas que garantizan la integridad de los datos dentro de la base de datos. Las restricciones comunes incluyen:
   - **NO NULO**: Asegura que una columna no pueda tener un valor nulo.
   - **ÚNICO**: Asegura que todos los valores en una columna sean distintos.
   - **VERIFICAR (CHECK)**: Asegura que los valores en una columna cumplan criterios específicos.
   - **DEFECTO (DEFAULT)**: Proporciona un valor predeterminado para una columna si no se especifica un valor durante la inserción.

## <font color='8EC044'> **Definición `SQL`** </font>

El lenguaje de consulta estructurada (`SQL`, Structured Query Language) es un lenguaje de programación para almacenar y procesar información en una base de datos relacional.

* **Estructurada** (*Structured*): Los datos están estructurados. Trabajas con campos *fijos*.
*  **Consulta** (*Query*): Solicitas información usando una consulta de texto.
* **Lenguaje** (*Language*): SQL es otro lenguaje de programación




## <font color='8EC044'> **`Pandas` o `SQL`** </font>

`SQL` es un lenguaje de consulta estructurado diseñado específicamente para interactuar con bases de datos relacionales, mientras que `Pandas` es una libreria de `Python` que proporciona estructuras y funciones para manipular y analizar datos en memoria.

La diferencia fundamental radica en que `SQL` es un lenguaje independiente que se utiliza para describir operaciones sobre datos almacenados en sistemas de gestión de bases de datos, mientras que `Pandas` es una herramienta dentro del ecosistema de Python que permite realizar tareas similares —como filtrado, agrupación y transformación— pero desde un entorno de programación general. Así, SQL se enfoca en el **qué** se quiere hacer con los datos, mientras que Pandas ofrece más control sobre el **cómo**, integrándose fácilmente en flujos de análisis y automatización más amplios.


Tanto `SQL` como `Pandas` pueden procesar datos. `SQL` es un lenguaje de operación de bases de datos que ha existido durante casi 50 años, mientras que  `Pandas` es una biblioteca de análisis/procesamiento de datos que fue escrita en Python en 2008.
* Algunos profesionales de datos (es decir, analistas de datos, ingenieros de datos) prefieren usar solo `SQL` para procesar datos
* Otros profesionales de datos (es decir, científicos de datos) prefieren usar `Pandas` para procesar datos y `SQL` solo para recuperar datos.

Aunque son herramientas de proposito distinto, con algunas funcionalidades en común, La pregunta que usualmente surge es cuando debo usar una o la otra.

### <font color='157699'> **Respecto al desempeño** </font>

Al considerar el volumen de los datos, la elección entre `SQL` y `Pandas` puede tener un impacto significativo en el rendimiento y la eficiencia. Estos son algunos aspectos claves a tener en cuenta:

* **Grandes volúmenes de datos**

  `SQL` está específicamente diseñado para gestionar grandes volúmenes de datos de manera eficiente. Puede manejar miles de millones de filas sin necesidad de cargar todos los datos en la memoria, lo que lo hace adecuado para bases de datos extensas.  `Pandas` es excelente para conjuntos de datos pequeños o subconjuntos de datos más grandes (por ejemplo, después de un filtrado inicial en SQL)



*  **Eficiencia de red**

   `SQL` permite realizar operaciones directamente en el servidor de la base de datos, minimizando la cantidad de datos transferidos a través de la red. `Pandas` opera en memoria, lo que significa que puede tener problemas con conjuntos de datos muy grandes que excedan la RAM de tu sistema.

* **Consultas complejas**:

  Para operaciones que involucran múltiples tablas o combinaciones complejas, `SQL` ejecuta estas consultas en el lado del servidor, lo que es más eficiente que cargar todos los datos relevantes en la memoria para manipularlos en Pandas.

## <font color='8EC044'> **Sistemas de administración de bases de datos relacionales** </font>

Un sistema de gestión de bases de datos relacional (*RDBMS*) es una aplicación de software que permite a los usuarios crear, almacenar y gestionar bases de datos relacionales.


Los *RDBMS* más comunes incluyen:

* **MySQL**: Un sistema de gestión de bases de datos de código abierto muy popular, ampliamente utilizado en aplicaciones web y proyectos de desarrollo.

* **PostgreSQL**: Otro sistema de gestión de bases de datos de código abierto que se destaca por su conformidad con los estándares SQL y su capacidad para manejar consultas complejas.

* **Oracle Database**: Un RDBMS comercial conocido por su robustez, escalabilidad y soporte para aplicaciones empresariales.

* **Microsoft SQL Server**: Una solución de base de datos de Microsoft que se integra bien con otros productos de Microsoft y es popular en entornos empresariales.

* **SQLite**: Un sistema de gestión de bases de datos ligero y de código abierto, que se utiliza a menudo en aplicaciones móviles y en el almacenamiento local debido a su bajo consumo de recursos.

* **MariaDB**: Un fork de MySQL, desarrollado por los creadores originales de MySQL, que busca ofrecer una alternativa completamente de código abierto.

* **IBM Db2**: Un sistema de gestión de bases de datos comercial que es conocido por su alto rendimiento y capacidad de escalado, utilizado en grandes empresas.

## <font color='8EC044'> **Crear tablas** </font>


Se crearán dos tablas denominadas `fabricante` y `poducto`.

```
ALTER TABLE fabricante
ADD COLUMN pais VARCHAR(100);
```



## <font color='8EC044'> **`DML` – Lenguaje de Manipulación de Datos** </font>
Se utiliza para manipular los datos dentro de las tablas, incluyendo operaciones como insertar, actualizar y eliminar registros. Algunas clausulas comunes son
  * `INSERT`: Agregar nuevos registros a una tabla.
  * `UPDATE`: Modificar registros existentes en una tabla.
  * `DELETE`: Eliminar registros de una tabla.

  La sintaxis es la siguiente:

| Command | Syntax |
|---------|--------|
| INSERT  | `INSERT INTO table_name (column1, column2, ...) VALUES (value1, value2, ...);` |
| UPDATE  | `UPDATE table_name SET column1 = value1, column2 = value2 WHERE condition;`    |
| DELETE  | `DELETE FROM table_name WHERE condition;`                                      |

Con base en la columna país que creamos podriamos agregar los datos correspondientes



```

SELECT producto.nombre
FROM producto
JOIN fabricante ON producto.id_fabricante = fabricante.id
WHERE fabricante.pais = 'China';
```







## <font color='8EC044'> **`DCL` – Lenguaje de Control de Datos** </font>
Se utiliza para controlar el acceso a los datos en la base de datos, permitiendo otorgar o revocar permisos a los usuarios. Siendo comandos comunes:
  * `GRANT`: Proporcionar privilegios específicos a usuarios o roles.
  * `REVOKE`: Eliminar privilegios previamente otorgados.

  La sintaxis de las instrucciones es la siguiente:

| Command | Syntax |
|---------|--------|
| GRANT   | `GRANT privilege_type [(column_list)] ON [object_type] object_name TO user [WITH GRANT OPTION];` |
| REVOKE  | `REVOKE [GRANT OPTION FOR] privilege_type [(column_list)] ON [object_type] object_name FROM user [CASCADE];` |



## <font color='8EC044'> **`TCL` – Lenguaje de Control de Transacciones** </font>

Se utiliza para manejar las transacciones en la base de datos, asegurando que las operaciones sean completadas de manera segura y coherente. Incluye clausulas como:
  * `COMMIT`: Guardar todos los cambios realizados durante la transacción actual.
  * `ROLLBACK`: Deshacer los cambios realizados durante la transacción actual si ocurre un error.
  * `SAVEPOINT`: Crear un punto dentro de una transacción al cual se puede volver más tarde.

  La sintaxis es la siguiente:

| Command           | Syntax                                           |
|-------------------|--------------------------------------------------|
| COMMIT            | `COMMIT;`                                        |
| ROLLBACK          | `ROLLBACK;`                                      |
| SAVEPOINT         | `SAVEPOINT savepoint_name;`                      |


## <font color='46B8A9'> **Ejercicio** </font>


Replique las operaciones descritas en `SQLlite`

# <font color='056938'> **Operadores** </font>

al igual que en otros lenguajes como `python`,  los operadores en `SQL` son símbolos o palabras clave que permiten realizar operaciones sobre los datos en una base de datos. Se utilizan para manipular, comparar, filtrar y transformar la información dentro de las consultas SQL. Estos se clasifican segun su tipo, algunos de ellos son:



## <font color='8EC044'> **Operadores Aritméticos** </font>

| Operador | Descripción            |
|----------|------------------------|
| `+`      | Suma                   |
| `-`      | Resta                  |
| `*`      | Multiplicación          |
| `/`      | División               |
| `%`      | Módulo (resto de división) |




## <font color='8EC044'> **Operadores de Comparación** </font>

| Operador  | Descripción               |
|-----------|---------------------------|
| `=`       | Igual                     |
| `!=` o `<>` | Distinto                  |
| `>`       | Mayor que                 |
| `<`       | Menor que                 |
| `>=`      | Mayor o igual que         |
| `<=`      | Menor o igual que         |


## <font color='8EC044'> **Operadores Lógicos** </font>

| Operador | Descripción                                     |
|----------|-------------------------------------------------|
| `AND`    | Verdadero si ambas condiciones son verdaderas   |
| `OR`     | Verdadero si al menos una condición es verdadera|
| `NOT`    | Invierte el resultado de una condición          |

A nivel de bits los operadores lógicos se expresan así:


| Operador | Descripción         |
|----------|---------------------|
| `&`      | AND a nivel de bits |
| `|`      | OR a nivel de bits  |
| `^`      | XOR a nivel de bits |
| `~`      | NOT a nivel de bits |
| `<<`     | Desplazamiento a la izquierda |
| `>>`     | Desplazamiento a la derecha   |



## <font color='8EC044'> **Operadores de Conjuntos** </font>


| Operador   | Descripción                                               |
|------------|-----------------------------------------------------------|
| `UNION`    | Combina conjuntos de resultados y elimina duplicados       |
| `UNION ALL`| Combina conjuntos de resultados sin eliminar duplicados    |
| `INTERSECT`| Devuelve registros comunes entre ambos conjuntos de resultados |
| `EXCEPT`   | Devuelve los registros del primer conjunto que no están en el segundo |



## <font color='8EC044'> **Operadores de Cadenas** </font>

| Operador | Descripción                                                                 |
|----------|-----------------------------------------------------------------------------|
| `||` o `+` | Concatenación                                                             |
| `LIKE`   | Se usa con comodines (`%` para múltiples caracteres, `_` para un solo carácter) |
| `IN`     | Verifica si un valor coincide con algún valor en una lista                  |
| `BETWEEN`| Verifica si un valor está dentro de un rango                                |




## <font color='8EC044'> **Otros Operadores** </font>

| Operador  | Descripción                                        |
|-----------|----------------------------------------------------|
| `IS NULL` | Verifica si un valor es `NULL`                     |
| `EXISTS`  | Verifica si una subconsulta devuelve algún resultado|
| `ALL`     | Compara un valor con todos los valores de una subconsulta |
| `ANY` o `SOME` | Compara un valor con cualquier valor de una subconsulta |


# <font color='056938'> **Sentencias o `queries`** </font>

En SQL, las clausulas y los operadores son usados  para construir **sentencias**, usualamente llamadas *Query*. Estas son instrucciones completas que se utilizan para realizar acciones específicas en una base de datos. Cada sentencia puede incluir varias cláusulas y operadores que especifican las condiciones, filtros o acciones adicionales que se deben aplicar.

Por ejemplo en la sentencia

```sql
SELECT nombre, precio
FROM producto
WHERE precio > 150
ORDER BY nombre DESC;
```


- **`SELECT`**: Esta cláusula especifica qué columnas se desean recuperar (en este caso, `nombre` y `precio`).
- **`FROM`**: Esta cláusula indica de qué tabla se obtendrán los datos (en este caso, de la tabla `producto`).
- **`WHERE`**: Esta cláusula se utiliza para filtrar los resultados y solo incluir aquellos productos cuyo precio sea mayor a 1500.
- **`ORDER BY`**: Esta cláusula se utiliza para ordenar los resultados en orden descendente según el nombre.



Aunque no todas las consultas incluyen todas las cláusulas, cada parte de la sentencia SQL se ejecuta en una secuencia determinada:

* `FROM` y `JOIN`, para determinar las fuentes de los datos consultados
* `WHERE`, para incluir sólo los datos que cumplan criterios específicos
* `GROUP BY`, para resumir datos basándose en valores comunes.
* `HAVING`, para incluir sólo los datos de las Filas agregadas que cumplan unos criterios específicos.
* `SELECT` o `SELECT DISTINCT`, para especificar aún más qué datos de origen incluir
* `ORDER BY`, para especificar cómo ordenar los datos en los resultados.
* `LIMIT y OFFSET`, para especificar cuántos -y qué- registros de la consulta incluir.

# <font color='056938'> **Uniendo tablas** </font>

Es comun que necesitamos hacer preguntas más complejas que requieren simultáneamente datos de dos tablas diferentes. Para ello usamos `JOIN`. Algunos de los tipos de JOIN se describen en la siguiente tabla.  


| Tipo de JOIN      | Descripción                                      | Devuelve                                      |
|-------------------|--------------------------------------------------|-----------------------------------------------|
| **INNER JOIN**    | Solo filas coincidentes de ambas tablas         | Filas con coincidencias en ambas tablas      |
| **LEFT JOIN**     | Todas las filas de la tabla izquierda y coincidentes de la derecha | Todas las filas de la izquierda, filas coincidentes de la derecha |
| **RIGHT JOIN**    | Todas las filas de la tabla derecha y coincidentes de la izquierda | Todas las filas de la derecha, filas coincidentes de la izquierda |
| **FULL JOIN**     | Todas las filas de ambas tablas                  | Todas las filas de ambas, NULL donde no hay coincidencia |
| **CROSS JOIN**    | Producto cartesiano de ambas tablas               | Cada combinación de filas de ambas tablas     |
| **SELF JOIN**     | Unión de una tabla consigo misma                 | Filas de la misma tabla según una condición   |

Quiza el tipo más comun es el `INNER JOIN`, de hecho cuando aplicamos `JOIN` sin especificar su tipo, `SQL` asume por defecto el tipo `INNER`



La forma más común de unir tablas es utilizando columnas de clave primaria y clave foránea.



> **Clave Primaria**:  
un identificador único para cada fila en una tabla.  

> **Clave Foránea**:  
se utiliza para hacer referencia a datos en una tabla desde otra tabla.


Por ejemplo, la siguiente instrucción

```sql
SELECT producto.nombre, fabricante.nombre
FROM producto
JOIN fabricante ON producto.id_fabricante = fabricante.id;
```
