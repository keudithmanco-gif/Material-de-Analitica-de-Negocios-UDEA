## <font color='8EC044'> **Conceptos básicos de `HTML`** </font>

`HTML` (Hypertext Markup Language) es el lenguaje de marcado estándar utilizado para crear páginas web y aplicaciones web. `HTML` se utiliza para describir la estructura y el contenido de una página web. `HTML` utiliza **etiquetas** y **atributos** para definir y dar formato al contenido de una página web, como el texto, las imágenes, los enlaces, los formularios y otros elementos.



### <font color='46B8A9'> **Etiquetas** </font>

Una etiqueta HTML es un elemento de marcado utilizado para definir diferentes tipos de contenido en una página web. Las etiquetas HTML están compuestas por un nombre de etiqueta rodeado de corchetes angulares, como `<etiqueta>`.

Las etiquetas HTML son utilizadas para **estructurar el contenido de una página web**, como agregar encabezados, párrafos, imágenes, enlaces, tablas, formularios, entre otros elementos. También se utilizan para **aplicar estilos y formatos a los elementos**, como cambiar el tamaño de una imagen, el color de fondo de un elemento o el tipo de letra de un texto. Agunas de las etiquetas más usadas son:

| Etiqueta | Descripción |
| --- | --- |
| `<html>` | Define un documento HTML |
| `<head>` | Contiene información sobre el documento, como el título y los enlaces a hojas de estilo |
| `<title>` | Define el título del documento |
| `<body>` | Contiene el contenido del documento, como texto, imágenes, enlaces y otros elementos |
| `<h1>`, `<h2>`, `<h3>`, `<h4>`, `<h5>`, `<h6>` | Encabezados de diferentes niveles |
| `<p>` | Define un párrafo |
| `<a>` | Define un enlace |
| `<img>` | Inserta una imagen en la página web |
| `<ul>`, `<ol>`, `<li>` | Crea listas desordenadas y ordenadas |
| `<table>`, `<tr>`, `<td>` | Crea tablas |
| `<form>`, `<input>`, `<button>` | Crea formularios y elementos de entrada, como campos de texto y botones |
| `<div>`, `<span>` | Crea secciones o bloques de contenido para aplicar estilos o scripts |
| `<style>`, `<script>` | Inserta estilos CSS y scripts JavaScript en la página web |


Este sería un ejemplo básico de elementos de una página web en `html` identificados con etiquetas serían:

```html
<!DOCTYPE html>
<html>
  <head>
    <title>Este es el título de la página</title>
  </head>
  <body>
    <!-- Aquí iría el contenido de la página -->
  </body>
</html>
```




### <font color='46B8A9'> **Atributos** </font>

En HTML, un atributo es una característica adicional que se agrega a una etiqueta para proporcionar información adicional sobre el elemento. Los atributos se utilizan para cambiar el comportamiento de las etiquetas y personalizar el contenido de la página web. Cada etiqueta HTML tiene sus propios atributos específicos, que determinan cómo se visualiza o se comporta el elemento en la página web.

Los atributos se agregan a las etiquetas mediante el uso de la sintaxis de nombre-valor. El nombre del atributo se coloca dentro de la etiqueta, y su valor se coloca dentro de comillas, como en el siguiente ejemplo:

```html
<img src="imagen.jpg" alt="Descripción de la imagen">
```

En este ejemplo, el atributo `src` se utiliza para especificar la ubicación de la imagen que se va a mostrar en la página web, mientras que el atributo `alt` proporciona una descripción alternativa de la imagen en caso de que no se pueda mostrar.

Los atributos más frecuentemente usados son

| Atributo | Definición |
| --- | --- |
| `href` | Especifica la URL de destino para un enlace |
| `src` | Especifica la fuente de una imagen o de otro elemento multimedia |
| `alt` | Proporciona un texto alternativo para una imagen o para otro elemento multimedia |
| `style` | Especifica un estilo CSS para un elemento |
| `class` | Especifica una o varias clases CSS para un elemento |
| `id` | Especifica un identificador único (ID) para un elemento |
| `title` | Proporciona información adicional sobre un elemento |
| `target` | Especifica en qué ventana o marco abrir el enlace |
| `rel` | Especifica la relación entre la página actual y la página vinculada |
| `type` | Especifica el tipo de un elemento |
| `name` | Especifica un nombre para un elemento |
| `value` | Especifica el valor de un elemento |
| `checked` | Indica que un elemento de entrada debe estar preseleccionado |
| `disabled` | Indica que un elemento de entrada está deshabilitado |
| `selected` | Indica que un elemento de opción debe estar preseleccionado |

### <font color='46B8A9'> **Obtener información estructurada desde la `url`** </font>

Importamos las librerias requeridas, en este caso lo haremos con `BeautifulSoup` y requerimos además la libreria `request` para obtener el contenido de la página.

Obtengamos los bloques de información de las bicicletas en la página, los cuales como ya hemos dicho están identificadas con el marcador:

```html
<div class="card-product relative h-full>
```

````python
from bs4 import BeautifulSoup
import requests as rq
import re

# Hacemos un request para obtener todo el contenido de la pagina
url = 'https://velominata.co/collections/bicicletas'
page = rq.get(url)
page.encoding = page.apparent_encoding

# convertimos el contenido en un objeto soup
soup = BeautifulSoup(page.content, 'html.parser')

# Obtenemos un listado de todos los bloques de información de las bicicletas
all_cards = soup.find_all('div', {'class': 'card-product relative h-full'})
all_cards

````

### <font color='46B8A9'> **Iterar en el proceso** </font>

Ahora, iteraremos en los pasos descritos anteriormente para extraer la siguiente información de cada vehículo publicado la cual almacenaremos en las siguientes listas:
* `enlaces`: dirección url del anuncio
* `precios_originales`: precio original
* `precios_rebajas`: precio rebaja
* `descripciones`: descripción de la bicicleta


````python
from bs4 import BeautifulSoup
import requests as rq
import re


# Crear listas vacías para cada una de las columnas:
enlaces = []
precios_originales = []
precios_rebajas = []
descripciones = []

# Hacemos un request para obtener todo el contenido de la pagina
url = 'https://velominata.co/collections/bicicletas'
page = rq.get(url)
page.encoding = page.apparent_encoding

# convertimos el contenido en un objeto soup
soup = BeautifulSoup(page.content, 'html.parser')

# Obtenemos un listado de todos los bloques de información de las bicicletas
all_cards = soup.find_all('div', {'class': 'card-product relative h-full'})

for card in all_cards:
  # enlace
  a_tag = card.find('a')
  enlaces.append(a_tag["href"])
  # Precio original
  precio_original_span  = card.find("span", class_="text-sm font-normal line-through text-neutral-500")
  precio_original = precio_original_span.text
  precio_original = precio_original.replace("$", "").replace(".", "")
  precios_originales.append(int(precio_original))
  # Precio rebaja
  precio_rebaja_span = card.find("span", class_="font-semibold text-base text-velo-green-light")
  precio_rebaja = precio_rebaja_span.text
  precio_rebaja = precio_rebaja.replace("$", "").replace(".", "")
  precios_rebajas.append(int(precio_rebaja))
  # Descripcion
  tag_descripcion = card.find("h3", class_="mb-4 basis-full text-base font-semibold")
  descripciones.append(tag_descripcion.text)
````

Convirtamos la información a un `DataFrame`

````python
import pandas as pd
df_info = pd.DataFrame({"enlaces": enlaces,
                        "precios_originales": precios_originales,
                        "precios_rebajas": precios_rebajas,
                        "descripciones": descripciones})
df_info
````

## <font color='8EC044'> **Usando `Selenium`** </font>

`BeautifulSoup` es una librería ligera, fácil de usar enfocada únicamente en analizar y extraer información de código HTML estático, lo que entre otras limitaciones significa que no puede ejecutar las rutinas de `JavaScript`. Por su parte, `Selenium` es una herramienta diseñada para automatizar navegadores web. Permite abrir páginas, simular clics, escribir en formularios y, sobre todo, esperar a que el contenido dinámico generado con JavaScript se cargue por completo. De esta manera, Selenium actúa como un usuario real que interactúa con la página, lo que lo hace ideal para obtener datos de sitios que no muestran toda la información en el HTML inicial.



No ahondaremos en el uso de Selenium pero evidenciaremos como su uso es bastante similar al de las démas librerias. En particular usaremos `google-colab-selenium`, un paquete que facilita usar Selenium dentro de Google Colab, ya que configura automáticamente `Chrome` y `Chromedriver` en el entorno de Colab, lo cual suele ser una tarea complicada

````python
%pip install -q google-colab-selenium
import google_colab_selenium as gs
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

# Listas vacías
enlaces = []
precios_originales = []
precios_rebajas = []
descripciones = []

# Iniciar Chrome
driver = gs.Chrome()
driver.get("https://velominata.co/collections/bicicletas")

# Esperar que carguen las cards
cards = WebDriverWait(driver, 15).until(
    EC.presence_of_all_elements_located((By.CSS_SELECTOR, "div.card-product.relative.h-full"))
)

for card in cards:
    # Enlace
    try:
        a_tag = card.find_element(By.TAG_NAME, "a")
        enlaces.append(a_tag.get_attribute("href"))
    except:
        enlaces.append(None)

    # Precio original (tachado)
    try:
        precio_original_span = card.find_element(By.CSS_SELECTOR, "span.text-sm.font-normal.line-through.text-neutral-500")
        precio_original = precio_original_span.text.replace("$", "").replace(".", "")
        precios_originales.append(int(precio_original))
    except:
        precios_originales.append(None)

    # Precio con rebaja
    try:
        precio_rebaja_span = card.find_element(By.CSS_SELECTOR, "span.font-semibold.text-base.text-velo-green-light")
        precio_rebaja = precio_rebaja_span.text.replace("$", "").replace(".", "")
        precios_rebajas.append(int(precio_rebaja))
    except:
        precios_rebajas.append(None)

    # Descripción
    try:
        tag_descripcion = card.find_element(By.CSS_SELECTOR, "h3.mb-4.basis-full.text-base.font-semibold")
        descripciones.append(tag_descripcion.text)
    except:
        descripciones.append(None)

driver.quit()


````

Convirtamos la información a un `DataFrame`

````python
import pandas as pd
df_info = pd.DataFrame({"enlaces": enlaces,
                        "precios_originales": precios_originales,
                        "precios_rebajas": precios_rebajas,
                        "descripciones": descripciones})
df_info
````
