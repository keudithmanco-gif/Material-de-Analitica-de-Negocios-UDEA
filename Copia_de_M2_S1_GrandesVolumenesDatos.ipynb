# <font color='056938'> **Grandes volúmenes de datos - </font> <font color='0F6CBF'>  Big Data** </font>
---

El término "Big Data" se refiere a conjuntos de datos extremadamente grandes y complejos que superan las capacidades de las herramientas y métodos tradicionales de procesamiento y análisis de datos. Estos conjuntos de datos se caracterizan por tres aspectos clave conocidos como las "3V": Volumen, Velocidad y Variedad.



1. **Volumen:** Big Data implica la gestión y análisis de cantidades masivas de datos. Estos datos pueden provenir de diversas fuentes, como redes sociales, sensores, transacciones comerciales y más. El volumen de datos es tan grande que no se pueden manejar fácilmente con herramientas convencionales de procesamiento y almacenamiento.

La menor unidad básica de medida es el `bit`, representado por un uno o un cero. Sin embargo, los bits se agregan en `bytes`, ocho bits corresponden a un byte. El byte es la unidad de almacenamiento requerida por ejemplo para guardar la letra a. Los volúmenes de datos se miden en unidades derivadas del byte

| Unidad      | Símbolo | Equivalencia                      |
|-------------|---------|----------------------------------|
| Byte        | B       | 1 byte                           |
| Kilobyte    | KB      | 1,024 bytes                      |
| Megabyte    | MB      | 1,024 KB = 1,048,576 bytes       |
| Gigabyte    | GB      | 1,024 MB = 1,073,741,824 bytes   |
| Terabyte    | TB      | 1,024 GB = 1,099,511,627,776 bytes|
| Petabyte    | PB      | 1,024 TB = 1,125,899,906,842,624 bytes|
| Exabyte     | EB      | 1,024 PB = 1,152,921,504,606,846,976 bytes|
| Zettabyte   | ZB      | 1,024 EB = 1,180,591,620,717,411,303,424 bytes|
| Yottabyte   | YB      | 1,024 ZB = 1,208,925,819,614,629,174,706,176 bytes|




2. **Velocidad:** Los datos en el mundo actual se generan a una velocidad asombrosa. Big Data se refiere a la capacidad de manejar y analizar estos datos en tiempo real o cerca de tiempo real, lo que requiere sistemas y herramientas que puedan procesar y actuar sobre la información rápidamente.

La velocidad con la que se generan los datos se mide en términos de "bits por segundo" (bps) o múltiplos de bits por segundo


| Unidad                          | Equivalencia           |
|---------------------------------|------------------------|
| Bits por segundo (bps)          | 1 bit por segundo      |
| Kilobits por segundo (Kbps)     | 1,000 bps (10^3 bps)   |
| Megabits por segundo (Mbps)     | 1,000 Kbps (10^6 bps)  |
| Gigabits por segundo (Gbps)     | 1,000 Mbps (10^9 bps)  |
| Terabits por segundo (Tbps)     | 1,000 Gbps (10^12 bps) |





3. **Variedad:** Los datos en el mundo de Big Data no solo son numéricos, sino que también incluyen texto, imágenes, videos, audio y otros formatos. La diversidad de fuentes y tipos de datos es una característica esencial del Big Data.




En algunos casos, se han agregado más "V" para abordar otros desafíos, como la Veracidad y la Variabilidad. Aquí está la explicación de las principales características:

4. **Veracidad:** A medida que la cantidad de datos aumenta, también lo hace la posibilidad de que algunos de los datos sean incorrectos, engañosos o inexactos. Mantener la veracidad de los datos es un desafío importante en el análisis de Big Data.

5. **Variabilidad:** Los datos pueden tener una estructura variable, lo que significa que no todos los datos se almacenan en un formato uniforme. Esto agrega complejidad al proceso de análisis y requiere herramientas y enfoques flexibles.

6. **Valor:** El propósito final de trabajar con Big Data es extraer información valiosa, patrones y conocimientos que puedan generar beneficios para las organizaciones y la toma de decisiones.

# <font color='056938'> **Fuentes de datos** </font>

En análisis, las fuentes de datos se refieren a los orígenes o lugares desde los cuales se recopila, procesa y analiza información para obtener ideas y tomar decisiones informadas. Algunas fuentes de datos comunes utilizadas en análisis son:

1. **Sistemas Transaccionales**: Estas son bases de datos o sistemas que registran transacciones comerciales cotidianas, como ventas, pedidos, facturas e interacciones con clientes. Algunos ejemplos incluyen sistemas de Punto de Venta (POS), sistemas de Planificación de Recursos Empresariales (ERP) y sistemas de Gestión de Relaciones con Clientes (CRM).

2. **Aplicaciones Web y Móviles**: Los datos de sitios web y aplicaciones móviles pueden proporcionar información sobre el comportamiento del usuario, interacciones, clics, vistas de páginas, descargas y más. Estos datos se recopilan mediante herramientas como Google Analytics, Adobe Analytics o soluciones de seguimiento personalizadas.

3. **Plataformas de Redes Sociales**: Los datos de redes sociales incluyen métricas como "Me gusta", compartidos, comentarios y participación del usuario. Se pueden utilizar APIs proporcionadas por plataformas como Facebook, Twitter, Instagram y LinkedIn para recopilar estos datos.

4. **Datos de Sensores**: En campos como Internet de las Cosas (IoT) y fabricación, los datos de sensores de dispositivos, máquinas y equipos pueden proporcionar información sobre el rendimiento, patrones de uso y anomalías.

5. **Encuestas y Formularios de Retroalimentación**: Las encuestas, cuestionarios y formularios de retroalimentación proporcionan datos cualitativos y cuantitativos de clientes, usuarios o empleados, ayudando a las organizaciones a comprender opiniones, preferencias y necesidades.

6. **Datos de Investigación de Mercado**: Los datos recopilados por firmas externas de investigación de mercado pueden incluir tendencias industriales, análisis de competidores, segmentación de clientes e información demográfica.

7. **Datos Financieros**: Los estados financieros, registros de transacciones e informes financieros brindan información sobre la salud financiera, el rendimiento y las tendencias de una organización.

8. **Archivos de Registro**: Los registros de servidores, registros de aplicaciones y registros de eventos contienen información valiosa sobre el rendimiento del sistema, errores, interacciones del usuario y patrones de uso.

9. **Proveedores Externos de Datos**: Muchas organizaciones compran o se suscriben a servicios de datos que proporcionan datos específicos de la industria o demográficos. Ejemplos incluyen datos demográficos de la Oficina del Censo de EE. UU. o datos de mercado de proveedores como Nielsen.

10. **Bases de Datos Gubernamentales**: Las agencias gubernamentales a menudo proporcionan conjuntos de datos abiertos que se pueden utilizar para el análisis. Ejemplos incluyen datos de crímenes, datos ambientales e indicadores económicos.

11. **Datos de Soporte al Cliente y Servicio**: Los datos de interacciones de soporte al cliente, centros de llamadas y sistemas de ayuda pueden proporcionar información sobre la satisfacción del cliente, problemas comunes y eficiencia del soporte.

12. **Almacenes de Datos Internos y Lagos de Datos**: Las organizaciones pueden centralizar sus datos en almacenes de datos (`data warehose`) o lagos de datos (`datalakes`), integrando datos de diversas fuentes para análisis e informes.

13. **APIs de Terceros**: Las APIs proporcionadas por servicios de terceros permiten que los datos se extraigan directamente en herramientas de análisis. Por ejemplo, las APIs de pronóstico del tiempo pueden proporcionar datos históricos de clima para el análisis.

14. **Sistemas de Empleados y Recursos Humanos**: Los sistemas de RRHH contienen información sobre empleados, nómina, rendimiento, capacitación y más, que se pueden analizar para mejorar la gestión de la fuerza laboral.

# <font color='056938'> **Extracción Transformación y Carga -   ETL** </font>

`ETL`, por sus siglas en Ingles, significa extracción, transformación y carga. Es un proceso de integración de datos que combina información de múltiples fuentes en un conjunto de datos coherente que se carga en un almacén de datos (`data warehouse`) o en otro sistema de destino.

`ETL` proporciona la base para flujos de trabajo de análisis de datos y aprendizaje automático. A través de una serie de reglas del negocio, ETL limpia y organiza los datos de manera que aborde necesidades específicas. Las tres etapas del proceso usualmente comprenden:


## <font color='157699'> **Extracción** </font>

Durante la extracción de datos, los datos en bruto se copian o exportan desde ubicaciones fuente a un área de preparación. Los equipos de gestión de datos pueden extraer datos de una variedad de fuentes de datos, que pueden ser estructuradas o no estructuradas. Estas fuentes incluyen, entre otras:

* Servidores SQL o NoSQL
* Sistemas CRM y ERP
* Archivos planos
* Correos electrónicos
* Páginas web

## <font color='157699'> **Transformación** </font>
En el área de preparación, los datos en bruto son sometidos a procesamiento de datos. Aquí, los datos se transforman y consolidan para su caso de uso analítico previsto. Esta fase puede involucrar las  tareas como:

* Filtrar, limpiar, remover duplicados, validar y autenticar los datos.
* Realizar cálculos, traducciones o resumen o agrupaciones basadas en los datos en bruto.
* Realizar auditorías para garantizar la calidad de los datos
* Eliminar, cifrar o proteger datos regulados por regulaciones industriales o gubernamentales.
* Formatear los datos en tablas o tablas combinadas para que coincidan con el esquema del almacén de datos de destino.

## <font color='157699'> **Carga** </font>
En este último paso, los datos transformados se mueven desde el área de preparación a un almacén de datos de destino. Por lo general, esto implica una carga inicial de todos los datos, seguida de cargas periódicas de cambios incrementales en los datos y, menos frecuentemente, actualizaciones completas para borrar y reemplazar los datos en el almacén. Para la mayoría de las organizaciones que utilizan ETL, el proceso es automatizado, bien definido, continuo y basado en lotes.

`ETL` es solo una de varias posibles estrategias de integración de datos existentes. Así por ejemplo, dependiendo del volumen, la velocidad y el tipo de datos considerado existen otras estrategias como:

* **ELT** copia o exporta los datos desde las ubicaciones fuente, pero en lugar de cargarlos en un área de preparación para su transformación, carga los datos en bruto directamente en el almacenamiento de datos de destino para ser transformados según sea necesario.

* **Change Data Capture** (CDC) identifica y captura solo los datos fuente que han cambiado y traslada esos datos al sistema de destino. CDC se puede utilizar para reducir los recursos requeridos durante el paso de "extracción" en ETL

* La **virtualización** de datos utiliza una capa de abstracción de software para crear una vista unificada, integrada y completamente utilizable de los datos, sin copiar físicamente, transformar o cargar los datos fuente en un sistema de destino.

## <font color='157699'> **Servicios en la nube** </font>

Dada la dependencia que tiene el análisis de grandes volúmenes de datos en altas capacidades de cómputo, y los requerimientos que ello impone a las compañias, existen servicios que proveen de manera flexible dicha infraestructura.  Estos servicios usualmente se hacen accesibles a tráves de la red y hacen parte de los denominados servicios en la nube.


Las caracteríssticas básicas de este tipo de servicios son:

1. **Escalabilidad**: permite escalar recursos de cómputo, almacenamiento y redes de acuerdo a las necesidades de tu aplicación o negocio. Puedes aumentar o disminuir recursos de manera flexible.

2. **Variedad de Servicios**:  ofrece una gran cantidad de servicios, desde servidores virtuales (instancias EC2) hasta bases de datos gestionadas, herramientas de aprendizaje automático, almacenamiento en la nube y mucho más.

3. **Pago por Uso**: utiliza un modelo de precios de pago por uso, lo que significa que solo pagas por los recursos que consumes. Esto puede ser muy eficiente en términos de costos.

4. **Seguridad**: ofrece herramientas y servicios para proteger tus aplicaciones y datos, como cortafuegos, sistemas de autenticación y cifrado.

5. **Disponibilidad Global**:  opera en regiones y zonas de disponibilidad en todo el mundo, lo que te permite implementar aplicaciones en ubicaciones cercanas a tus usuarios para reducir la latencia.

Algunos de los usos comunes para este tipo de servicios son


- **Hospedaje de Sitios Web**: Permite alojar sitios web y aplicaciones web para asegurarte de que estén disponibles en todo momento y puedan manejar cargas de tráfico variables.

- **Almacenamiento en la Nube**:  Permite almacenar y acceder a datos de manera segura y escalable.

- **Desarrollo de Aplicaciones**: Puede usarse para construir, implementar y escalar aplicaciones de manera eficiente.

- **Análisis de Datos**:  Ofrece herramientas que permiten el procesamiento y análisis de grandes conjuntos de datos.

- **Aprendizaje Automático e Inteligencia Artificial**:  Proporcionan servicios y herramientas para el desarrollo y entrenamiento de modelos de aprendizaje automático.
